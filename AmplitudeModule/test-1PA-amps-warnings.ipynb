## Quasicular, equatorial restrictions

Check that the $\texttt{get\_amplitudes*}$ functions return error messages if called with $e \neq 0$ or $x_I \neq 1$.

```python
try:
    amp.get_amplitudes(0, 10.0, 0.1, 1.0, 1e-4, 0.0, 0.0, 0.0) #calling with e = 0.1
except Exception as error:
    print(f"{type(error).__name__}: {error}")
```

```python
try:
    amp.get_amplitudes(0, 10.0, 0.0, 0.9, 1e-4, 0.0, 0.0, 0.0) #calling with xI = 0.9
except Exception as error:
    print(f"{type(error).__name__}: {error}")
```

```python
try:
    amp.get_amplitudes_single_piece(0, 10.0, 0.1, 1.0, "0PA") #calling with e = 0.1
except Exception as error:
    print(f"{type(error).__name__}: {error}")
```

```python
try:
    amp.get_amplitudes_single_piece(0, 10.0, 0.0, 0.9, "0PA") #calling with xI = 0.9
except Exception as error:
    print(f"{type(error).__name__}: {error}")
```

## Physical bounds checking

Ensure that error messages are returned if $\texttt{get\_amplitudes}$ is evaluated at $(\nu, a, p, e, x)$ values outside the physical range of validity/range of the underlying dataset.

```python
# Evaluate at physically invalid mass ratio
try:
    amp.get_amplitudes(0, 10.0, 0.0, 1.0, 0.3, 0.0, 0.0, 0.0) #evaluating at nu = 0.3 -- unphysical
except Exception as error:
    print(f"{type(error).__name__}: {error}")
```

```python
# Evaluate at single value of p outside the grid range
try:
    amp.get_amplitudes(0, 100, 0, 1) #evaluating at p = 100 -- out of bounds
except Exception as error:
    print(f"{type(error).__name__}: {error}")
```

```python
# Batched evaluation on grid of p, some of which lies outside the grid range
try:
    amp.get_amplitudes(0, np.linspace(6.25, 100), 0, 1) #evaluating on 6.25 <= p <= 100M -- out of bounds
except Exception as error:
    print(f"{type(error).__name__}: {error}")
```
